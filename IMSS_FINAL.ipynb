```python
%%sql
PRAGMA foreign_keys = ON;

-- 1. Tabla: clinicas (9 columnas)
CREATE TABLE IF NOT EXISTS clinicas (
    id_clinica INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre_clinica VARCHAR(100) NOT NULL,
    nivel_atencion VARCHAR(50),
    calle_numero VARCHAR(150),
    colonia VARCHAR(100),
    ciudad VARCHAR(100),
    estado VARCHAR(50),
    codigo_postal VARCHAR(10),
    telefono_contacto VARCHAR(15)
);

-- 2. Tabla: departamentos (9 columnas)
CREATE TABLE IF NOT EXISTS departamentos (
    id_departamento INTEGER PRIMARY KEY AUTOINCREMENT,
    id_clinica INT,
    nombre_depto VARCHAR(100) NOT NULL,
    jefe_depto VARCHAR(100),
    extension_telefonica INT,
    ubicacion_piso VARCHAR(20),
    horario_apertura TIME,
    horario_cierre TIME,
    presupuesto_anual DECIMAL(12,2),
    FOREIGN KEY (id_clinica) REFERENCES clinicas(id_clinica)
);

-- 3. Tabla: medicos (9 columnas)
CREATE TABLE IF NOT EXISTS medicos (
    id_medico INTEGER PRIMARY KEY AUTOINCREMENT,
    id_departamento INT,
    matricula_imss VARCHAR(20) UNIQUE,
    nombre VARCHAR(50),
    apellido_paterno VARCHAR(50),
    apellido_materno VARCHAR(50),
    cedula_profesional VARCHAR(20),
    especialidad VARCHAR(80),
    turno_trabajo VARCHAR(20),
    FOREIGN KEY (id_departamento) REFERENCES departamentos(id_departamento)
);

-- 4. Tabla: pacientes (10 columnas)
CREATE TABLE IF NOT EXISTS pacientes (
    id_paciente INTEGER PRIMARY KEY AUTOINCREMENT,
    nss_imss VARCHAR(11) UNIQUE NOT NULL,
    curp VARCHAR(18) UNIQUE,
    nombre VARCHAR(50),
    apellido_paterno VARCHAR(50),
    apellido_materno VARCHAR(50),
    fecha_nacimiento DATE,
    genero VARCHAR(15),
    tipo_sangre VARCHAR(5),
    telefono VARCHAR(15)
);

-- 5. Tabla: citas_medicas (9 columnas)
CREATE TABLE IF NOT EXISTS citas_medicas (
    id_cita INTEGER PRIMARY KEY AUTOINCREMENT,
    id_paciente INT,
    id_medico INT,
    fecha_cita DATE,
    hora_cita TIME,
    numero_consultorio VARCHAR(10),
    estado_cita VARCHAR(20),
    motivo_visita VARCHAR(200),
    tipo_cita VARCHAR(30),
    FOREIGN KEY (id_paciente) REFERENCES pacientes(id_paciente),
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);

-- 6. Tabla: diagnosticos (8 columnas)
CREATE TABLE IF NOT EXISTS diagnosticos (
    id_diagnostico INTEGER PRIMARY KEY AUTOINCREMENT,
    id_cita INT,
    codigo_cie10 VARCHAR(10),
    descripcion_enfermedad VARCHAR(255),
    tipo_diagnostico VARCHAR(50),
    nivel_gravedad VARCHAR(20),
    observaciones_medico TEXT,
    fecha_registro DATE,
    FOREIGN KEY (id_cita) REFERENCES citas_medicas(id_cita)
);

-- 7. Tabla: medicamentos (9 columnas)
CREATE TABLE IF NOT EXISTS medicamentos (
    id_medicamento INTEGER PRIMARY KEY AUTOINCREMENT,
    clave_cuadro_basico VARCHAR(20) UNIQUE,
    nombre_generico VARCHAR(100),
    forma_farmaceutica VARCHAR(50),
    concentracion VARCHAR(50),
    laboratorio_fabricante VARCHAR(100),
    lote_actual VARCHAR(30),
    fecha_caducidad_lote DATE,
    existencia_piezas INT
);

-- 8. Tabla: recetas (8 columnas)
CREATE TABLE IF NOT EXISTS recetas (
    id_receta INTEGER PRIMARY KEY AUTOINCREMENT,
    id_cita INT,
    folio_receta VARCHAR(50) UNIQUE,
    fecha_emision DATE,
    fecha_limite_surtido DATE,
    estado_surtido VARCHAR(30),
    farmacia_entrega VARCHAR(100),
    firma_electronica VARCHAR(255),
    FOREIGN KEY (id_cita) REFERENCES citas_medicas(id_cita)
);

-- 9. Tabla: receta_detalle (8 columnas)
CREATE TABLE IF NOT EXISTS receta_detalle (
    id_detalle INTEGER PRIMARY KEY AUTOINCREMENT,
    id_receta INT,
    id_medicamento INT,
    cantidad_cajas INT,
    dosis_indicada VARCHAR(100),
    frecuencia_horas INT,
    duracion_dias INT,
    indicaciones_especiales TEXT,
    FOREIGN KEY (id_receta) REFERENCES recetas(id_receta),
    FOREIGN KEY (id_medicamento) REFERENCES medicamentos(id_medicamento)
);

-- 10. Tabla: incapacidades (9 columnas)
CREATE TABLE IF NOT EXISTS incapacidades (
    id_incapacidad INTEGER PRIMARY KEY AUTOINCREMENT,
    id_paciente INT,
    id_medico INT,
    folio_incapacidad VARCHAR(50) UNIQUE,
    tipo_riesgo VARCHAR(50),
    dias_autorizados INT,
    fecha_inicio DATE,
    fecha_fin DATE,
    estado_aprobacion VARCHAR(30),
    FOREIGN KEY (id_paciente) REFERENCES pacientes(id_paciente),
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);
```

```python
%%sql
-- Creación de la base de datos y tablas de prueba en SQLite
CREATE TABLE IF NOT EXISTS clinicas (
    id_clinica INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre_clinica TEXT NOT NULL,
    nivel_atencion TEXT NOT NULL,
    calle_numero TEXT NOT NULL,
    colonia TEXT NOT NULL,
    ciudad TEXT NOT NULL,
    estado TEXT NOT NULL,
    codigo_postal TEXT NOT NULL,
    telefono_contacto TEXT NOT NULL
);

CREATE TABLE IF NOT EXISTS departamentos (
    id_departamento INTEGER PRIMARY KEY AUTOINCREMENT,
    id_clinica INTEGER NOT NULL,
    nombre_depto TEXT NOT NULL,
    jefe_depto TEXT NOT NULL,
    extension_telefonica INTEGER NOT NULL,
    ubicacion_piso TEXT NOT NULL,
    horario_apertura TEXT NOT NULL,
    horario_cierre TEXT NOT NULL,
    presupuesto_anual REAL NOT NULL,
    FOREIGN KEY (id_clinica) REFERENCES clinicas(id_clinica)
);

CREATE TABLE IF NOT EXISTS medicos (
    id_medico INTEGER PRIMARY KEY AUTOINCREMENT,
    id_departamento INTEGER NOT NULL,
    matricula_imss TEXT UNIQUE NOT NULL,
    nombre TEXT NOT NULL,
    apellido_paterno TEXT NOT NULL,
    apellido_materno TEXT NOT NULL,
    cedula_profesional TEXT NOT NULL,
    especialidad TEXT NOT NULL,
    turno_trabajo TEXT NOT NULL,
    FOREIGN KEY (id_departamento) REFERENCES departamentos(id_departamento)
);

CREATE TABLE IF NOT EXISTS pacientes (
    id_paciente INTEGER PRIMARY KEY AUTOINCREMENT,
    nss_imss TEXT UNIQUE NOT NULL,
    curp TEXT UNIQUE NOT NULL,
    nombre TEXT NOT NULL,
    apellido_paterno TEXT NOT NULL,
    apellido_materno TEXT NOT NULL,
    fecha_nacimiento TEXT NOT NULL,
    genero TEXT NOT NULL,
    tipo_sangre TEXT NOT NULL,
    telefono TEXT NOT NULL
);

CREATE TABLE IF NOT EXISTS citas_medicas (
    id_cita INTEGER PRIMARY KEY AUTOINCREMENT,
    id_paciente INTEGER NOT NULL,
    id_medico INTEGER NOT NULL,
    fecha_cita TEXT NOT NULL,
    hora_cita TEXT NOT NULL,
    numero_consultorio TEXT NOT NULL,
    estado_cita TEXT NOT NULL,
    motivo_visita TEXT NOT NULL,
    tipo_cita TEXT NOT NULL,
    FOREIGN KEY (id_paciente) REFERENCES pacientes(id_paciente),
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);

CREATE TABLE IF NOT EXISTS diagnosticos (
    id_diagnostico INTEGER PRIMARY KEY AUTOINCREMENT,
    id_cita INTEGER NOT NULL,
    codigo_cie10 TEXT NOT NULL,
    descripcion_enfermedad TEXT NOT NULL,
    tipo_diagnostico TEXT NOT NULL,
    nivel_gravedad TEXT NOT NULL,
    observaciones_medico TEXT,
    fecha_registro TEXT NOT NULL,
    FOREIGN KEY (id_cita) REFERENCES citas_medicas(id_cita)
);

CREATE TABLE IF NOT EXISTS medicamentos (
    id_medicamento INTEGER PRIMARY KEY AUTOINCREMENT,
    clave_cuadro_basico TEXT UNIQUE NOT NULL,
    nombre_generico TEXT NOT NULL,
    forma_farmaceutica TEXT NOT NULL,
    concentracion TEXT NOT NULL,
    laboratorio_fabricante TEXT NOT NULL,
    lote_actual TEXT NOT NULL,
    fecha_caducidad_lote TEXT NOT NULL,
    existencia_piezas INTEGER NOT NULL
);

CREATE TABLE IF NOT EXISTS recetas (
    id_receta INTEGER PRIMARY KEY AUTOINCREMENT,
    id_cita INTEGER NOT NULL,
    folio_receta TEXT UNIQUE NOT NULL,
    fecha_emision TEXT NOT NULL,
    fecha_limite_surtido TEXT NOT NULL,
    estado_surtido TEXT NOT NULL,
    farmacia_entrega TEXT NOT NULL,
    firma_electronica TEXT NOT NULL,
    FOREIGN KEY (id_cita) REFERENCES citas_medicas(id_cita)
);

CREATE TABLE IF NOT EXISTS receta_detalle (
    id_detalle INTEGER PRIMARY KEY AUTOINCREMENT,
    id_receta INTEGER NOT NULL,
    id_medicamento INTEGER NOT NULL,
    cantidad_cajas INTEGER NOT NULL,
    dosis_indicada TEXT NOT NULL,
    frecuencia_horas INTEGER NOT NULL,
    duracion_dias INTEGER NOT NULL,
    indicaciones_especiales TEXT,
    FOREIGN KEY (id_receta) REFERENCES recetas(id_receta),
    FOREIGN KEY (id_medicamento) REFERENCES medicamentos(id_medicamento)
);

CREATE TABLE IF NOT EXISTS incapacidades (
    id_incapacidad INTEGER PRIMARY KEY AUTOINCREMENT,
    id_paciente INTEGER NOT NULL,
    id_medico INTEGER NOT NULL,
    folio_incapacidad TEXT UNIQUE NOT NULL,
    tipo_riesgo TEXT NOT NULL,
    dias_autorizados INTEGER NOT NULL,
    fecha_inicio TEXT NOT NULL,
    fecha_fin TEXT NOT NULL,
    estado_aprobacion TEXT NOT NULL,
    FOREIGN KEY (id_paciente) REFERENCES pacientes(id_paciente),
    FOREIGN KEY (id_medico) REFERENCES medicos(id_medico)
);
```

```python
%%sql
SELECT
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    c.fecha_cita AS "Fecha Cita",
    m.nombre || ' ' || m.apellido_paterno AS Medico,
    d.nombre_depto AS Departamento
FROM citas_medicas c
JOIN pacientes p ON c.id_paciente = p.id_paciente
JOIN medicos m ON c.id_medico = m.id_medico
JOIN departamentos d ON m.id_departamento = d.id_departamento;
```

```python
%%sql
SELECT
    d.id_diagnostico AS ID,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    d.codigo_cie10 AS "CIE-10",
    d.descripcion_enfermedad AS Diagnostico,
    d.tipo_diagnostico AS Tipo,
    d.nivel_gravedad AS Gravedad,
    d.observaciones_medico AS Observaciones,
    d.fecha_registro AS Fecha
FROM diagnosticos d
JOIN citas_medicas c ON d.id_cita = c.id_cita
JOIN pacientes p ON c.id_paciente = p.id_paciente;
```

```python
%%sql
SELECT
    d.id_departamento AS ID,
    c.nombre_clinica AS Clinica,
    d.nombre_depto AS Departamento,
    d.jefe_depto AS Jefe,
    d.extension_telefonica AS Ext,
    d.ubicacion_piso AS Piso,
    d.horario_apertura AS Apertura,
    d.horario_cierre AS Cierre,
    d.presupuesto_anual AS Presupuesto
FROM departamentos d
JOIN clinicas c ON d.id_clinica = c.id_clinica;
```

```python
%%sql
SELECT
    m.id_medico AS ID,
    m.matricula_imss AS Matricula,
    m.nombre || ' ' || m.apellido_paterno || ' ' || m.apellido_materno AS "Nombre Medico",
    m.especialidad AS Especialidad,
    d.nombre_depto AS Departamento,
    m.cedula_profesional AS Cedula,
    m.turno_trabajo AS Turno
FROM medicos m
JOIN departamentos d ON m.id_departamento = d.id_departamento;
```

```python
%%sql
SELECT
    id_paciente AS ID,
    nss_imss AS NSS,
    curp AS CURP,
    nombre || ' ' || apellido_paterno || ' ' || apellido_materno AS "Nombre Paciente",
    fecha_nacimiento AS Nacimiento,
    genero AS Genero,
    tipo_sangre AS Sangre,
    telefono AS Telefono
FROM pacientes;
```

```python
%%sql
SELECT
    r.id_receta AS ID,
    r.folio_receta AS Folio,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    r.fecha_emision AS Emision,
    r.fecha_limite_surtido AS "Limite Surtido",
    r.estado_surtido AS Estado,
    r.farmacia_entrega AS Farmacia,
    r.firma_electronica AS "Firma Digital"
FROM recetas r
JOIN citas_medicas c ON r.id_cita = c.id_cita
JOIN pacientes p ON c.id_paciente = p.id_paciente;
```

```python
%%sql
SELECT
    rd.id_detalle AS ID,
    r.folio_receta AS "Folio Receta",
    m.nombre_generico AS Medicamento,
    m.concentracion AS Concentracion,
    rd.cantidad_cajas AS Cajas,
    rd.dosis_indicada AS Dosis,
    rd.frecuencia_horas AS "Frecuencia (Hrs)",
    rd.duracion_dias AS "Duracion (Dias)"
FROM receta_detalle rd
JOIN recetas r ON rd.id_receta = r.id_receta
JOIN medicamentos m ON rd.id_medicamento = m.id_medicamento;
```

```python
%%sql
SELECT
    i.id_incapacidad AS ID,
    i.folio_incapacidad AS Folio,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    m.nombre || ' ' || m.apellido_paterno AS "Medico Expide",
    i.tipo_riesgo AS Riesgo,
    i.dias_autorizados AS "Dias Autorizados",
    i.fecha_inicio AS Inicio,
    i.fecha_fin AS Fin,
    i.estado_aprobacion AS Estado
FROM incapacidades i
JOIN pacientes p ON i.id_paciente = p.id_paciente
JOIN medicos m ON i.id_medico = m.id_medico;
```

```python
%%sql
SELECT
    c.id_cita AS Cita,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    m.nombre || ' ' || m.apellido_paterno AS Medico,
    c.fecha_cita AS Fecha,
    c.hora_cita AS Hora,
    c.estado_cita AS Estado
FROM citas_medicas c
INNER JOIN pacientes p ON c.id_paciente = p.id_paciente
INNER JOIN medicos m ON c.id_medico = m.id_medico;
```

```python
%%sql
SELECT
    p.nss_imss AS NSS,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    c.id_cita AS "ID Cita",
    COALESCE(c.estado_cita, 'Sin Citas Agendadas') AS "Estado Cita"
FROM pacientes p
LEFT JOIN citas_medicas c ON p.id_paciente = c.id_paciente;
```

```python
%%sql
SELECT
    d.nombre_depto AS Departamento,
    d.ubicacion_piso AS Piso,
    m.nombre || ' ' || m.apellido_paterno AS "Medico Asignado",
    COALESCE(m.especialidad, 'Puesto Vacante') AS Especialidad
FROM departamentos d
LEFT JOIN medicos m ON d.id_departamento = m.id_departamento;
```

```python
%%sql
SELECT
    r.folio_receta AS Folio,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    med.nombre_generico AS Medicamento,
    rd.cantidad_cajas AS Cajas,
    rd.dosis_indicada AS Dosis,
    r.estado_surtido AS "Estado Surtido"
FROM recetas r
INNER JOIN citas_medicas c ON r.id_cita = c.id_cita
INNER JOIN pacientes p ON c.id_paciente = p.id_paciente
INNER JOIN receta_detalle rd ON r.id_receta = rd.id_receta
INNER JOIN medicamentos med ON rd.id_medicamento = med.id_medicamento;
```

```python
%%sql
SELECT
    p.nss_imss AS NSS,
    p.nombre || ' ' || p.apellido_paterno AS Paciente,
    d.descripcion_enfermedad AS Diagnostico,
    COALESCE(i.folio_incapacidad, 'No Requiere') AS "Folio Incapacidad",
    COALESCE(i.dias_autorizados, 0) AS "Dias Licencia"
FROM citas_medicas c
INNER JOIN pacientes p ON c.id_paciente = p.id_paciente
INNER JOIN diagnosticos d ON c.id_cita = d.id_cita
LEFT JOIN incapacidades i ON p.id_paciente = i.id_paciente AND c.id_medico = i.id_medico;
```

```python
%%sql
SELECT
    c.nombre_clinica AS "Unidad Medica",
    d.nombre_depto AS "Departamento",
    COUNT(DISTINCT cm.id_cita) AS "Total Citas",
    COUNT(DISTINCT diag.id_diagnostico) AS "Total Diagnosticos",
    COUNT(DISTINCT r.id_receta) AS "Recetas Emitidas",
    COALESCE(SUM(inc.dias_autorizados), 0) AS "Total Dias Incapacidad"
FROM clinicas c
INNER JOIN departamentos d ON c.id_clinica = d.id_clinica
LEFT JOIN medicos m ON d.id_departamento = m.id_departamento
LEFT JOIN citas_medicas cm ON m.id_medico = cm.id_medico
LEFT JOIN diagnosticos diag ON cm.id_cita = diag.id_cita
LEFT JOIN recetas r ON cm.id_cita = r.id_cita
LEFT JOIN incapacidades inc ON m.id_medico = inc.id_medico
GROUP BY c.id_clinica, d.id_departamento
ORDER BY "Total Citas" DESC;
```
